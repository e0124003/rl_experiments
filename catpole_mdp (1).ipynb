```python
# CartPole MDP Exploration in Jupyter Notebook

# Install gymnasium library
!pip install gymnasium

# Import necessary libraries
import gymnasium as gym
import numpy as np

# Load the CartPole environment
env = gym.make('CartPole-v1')

# Reset the environment to get the initial state
state, info = env.reset()

# Display the action space and observation space
print(f"Action Space: {env.action_space}")
print(f"Observation Space: {env.observation_space}")

# Define MDP Components for CartPole

# State space
def describe_state(state):
    """
    This function prints out the individual components of the state
    State is a tuple (x, x_dot, theta, theta_dot)
    """
    cart_position, cart_velocity, pole_angle, pole_velocity = state
    print(f"Cart Position: {cart_position}")
    print(f"Cart Velocity: {cart_velocity}")
    print(f"Pole Angle: {pole_angle}")
    print(f"Pole Velocity: {pole_velocity}")

# Example of an initial state in CartPole
print("Initial State:")
describe_state(state)

# Action space exploration
# In CartPole, there are two actions: 0 (push left) and 1 (push right)
actions = {0: "Move Left", 1: "Move Right"}
for action in actions:
    print(f"Action {action}: {actions[action]}")

# Simulate a few steps to see state transitions and rewards
num_steps = 5
print("\nSimulating a few steps:")
for step in range(num_steps):
    action = env.action_space.sample()  # Random action
    next_state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated

    print(f"\nStep {step + 1}:")
    print(f"Action taken: {actions[action]}")
    print("Next State:")
    describe_state(next_state)
    print(f"Reward: {reward}")
    print(f"Done: {done}")

    if done:
        state, info = env.reset()
        print("Environment reset due to termination or truncation.")

# Close the environment when done
env.close()
```

```python
# Frozen Lake MDP Exploration in Jupyter Notebook
# Install gymnasium library
!pip install gymnasium

# Import necessary libraries
import gymnasium as gym
import numpy as np

# Load the FrozenLake environment (using standard 4x4 non-slippery for clear MDP analysis)
env = gym.make('FrozenLake-v1', is_slippery=False)

# Reset the environment to get the initial state
state, info = env.reset()

# Display the action space and observation space
print(f"Action Space: {env.action_space}")
print(f"Observation Space: {env.observation_space}")

# Define MDP Components for FrozenLake
# State space
def describe_state(state):
    """
    This function prints out the individual components of the state.
    For FrozenLake, the state is a single integer representing the
    agent's current grid position (0 to 15 on a 4x4 board).
    """
    print(f"Current State (Grid Index): {state}")

# Example of an initial state in FrozenLake
print("Initial State:")
describe_state(state)

# Action space exploration
# In FrozenLake, there are four discrete actions: 0 (Left), 1 (Down), 2 (Right), 3 (Up)
actions = {
    0: "Move Left",
    1: "Move Down",
    2: "Move Right",
    3: "Move Up"
}
for action in actions:
    print(f"Action {action}: {actions[action]}")

# Simulate a few steps to see state transitions and rewards
num_steps = 5
print("\nSimulating a few steps:")
for step in range(num_steps):
    action = env.action_space.sample()  # Random action selection
    next_state, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated

    print(f"\nStep {step + 1}:")
    print(f"Action taken: {actions[action]}")
    print("Next State:")
    describe_state(next_state)
    print(f"Reward: {reward}")
    print(f"Done: {done}")

    if done:
        state, info = env.reset()
        print("Environment reset due to termination or truncation.")

# Close the environment when done
env.close()
```

```python
# Task 6: Explore Mountain Car, Blackjack, Taxi, and Cliff Walking in OpenAI Gym
# Install gymnasium library
!pip install gymnasium

# Import necessary libraries
import gymnasium as gym
import numpy as np

# -------------------------------------------------------------
# 1. MOUNTAIN CAR ENVIRONMENT
# -------------------------------------------------------------
print("="*60)
print("1. EXPLORING MOUNTAIN CAR")
print("="*60)

env_mc = gym.make('MountainCar-v0')
state_mc, info_mc = env_mc.reset()

print(f"Action Space: {env_mc.action_space}")
print(f"Observation Space: {env_mc.observation_space}")

def describe_state_mc(state):
    """Prints out components of the Mountain Car state: [position, velocity]"""
    position, velocity = state
    print(f"Car Position: {position}")
    print(f"Car Velocity: {velocity}")

print("\nInitial State:")
describe_state_mc(state_mc)

# Actions: 0 (Accelerate Left), 1 (Don't accelerate), 2 (Accelerate Right)
actions_mc = {0: "Accelerate Left", 1: "Don't Accelerate", 2: "Accelerate Right"}
print("\nActions Available:")
for act in actions_mc:
    print(f"Action {act}: {actions_mc[act]}")

print("\nSimulating 2 steps:")
for step in range(2):
    action = env_mc.action_space.sample()
    next_state, reward, terminated, truncated, info = env_mc.step(action)
    done = terminated or truncated
    print(f"Step {step + 1}: Action taken -> {actions_mc[action]}")
    describe_state_mc(next_state)
    print(f"Reward: {reward} | Done: {done}")
env_mc.close()


# -------------------------------------------------------------
# 2. BLACKJACK ENVIRONMENT
# -------------------------------------------------------------
print("\n" + "="*60)
print("2. EXPLORING BLACKJACK")
print("="*60)

env_bj = gym.make('Blackjack-v1')
state_bj, info_bj = env_bj.reset()

print(f"Action Space: {env_bj.action_space}")
print(f"Observation Space: {env_bj.observation_space}")

def describe_state_bj(state):
    """Prints out components of Blackjack state: (player_sum, dealer_card, usable_ace)"""
    player_sum, dealer_card, usable_ace = state
    print(f"Player Current Hand Sum: {player_sum}")
    print(f"Dealer Showing Card: {dealer_card}")
    print(f"Has Usable Ace: {usable_ace}")

print("\nInitial State:")
describe_state_bj(state_bj)

# Actions: 0 (Stick/Stop), 1 (Hit/Take card)
actions_bj = {0: "Stick (Stop)", 1: "Hit (Take Card)"}
print("\nActions Available:")
for act in actions_bj:
    print(f"Action {act}: {actions_bj[act]}")

print("\nSimulating 2 steps:")
for step in range(2):
    action = env_bj.action_space.sample()
    next_state, reward, terminated, truncated, info = env_bj.step(action)
    done = terminated or truncated
    print(f"Step {step + 1}: Action taken -> {actions_bj[action]}")
    describe_state_bj(next_state)
    print(f"Reward: {reward} | Done: {done}")
    if done:
        print("Game finished. Resetting hand.")
        state_bj, info_bj = env_bj.reset()
env_bj.close()


# -------------------------------------------------------------
# 3. TAXI ENVIRONMENT
# -------------------------------------------------------------
print("\n" + "="*60)
print("3. EXPLORING TAXI")
print("="*60)

env_tx = gym.make('Taxi-v4') # Changed from 'Taxi-v3' to 'Taxi-v4'
state_tx, info_tx = env_tx.reset()

print(f"Action Space: {env_tx.action_space}")
print(f"Observation Space: {env_tx.observation_space}")

def describe_state_tx(state):
    """Prints out the single encoded scalar state for Taxi grid (0-499)"""
    print(f"Encoded Grid Position State Index: {state}")

print("\nInitial State:")
describe_state_tx(state_tx)

# Actions: 0 (South), 1 (North), 2 (East), 3 (West), 4 (Pickup), 5 (Dropoff)
actions_tx = {0: "Move South", 1: "Move North", 2: "Move East", 3: "Move West", 4: "Pickup Passenger", 5: "Dropoff Passenger"}
print("\nActions Available:")
for act in actions_tx:
    print(f"Action {act}: {actions_tx[act]}")

print("\nSimulating 2 steps:")
for step in range(2):
    action = env_tx.action_space.sample()
    next_state, reward, terminated, truncated, info = env_tx.step(action)
    done = terminated or truncated
    print(f"Step {step + 1}: Action taken -> {actions_tx[action]}")
    describe_state_tx(next_state)
    print(f"Reward: {reward} | Done: {done}")
env_tx.close()


# -------------------------------------------------------------
# 4. CLIFF WALKING ENVIRONMENT
# -------------------------------------------------------------
print("\n" + "="*60)
print("4. EXPLORING CLIFF WALKING")
print("="*60)

env_cw = gym.make('CliffWalking-v1') # Changed from 'CliffWalking-v0' to 'CliffWalking-v1'
state_cw, info_cw = env_cw.reset()

print(f"Action Space: {env_cw.action_space}")
print(f"Observation Space: {env_cw.observation_space}")

def describe_state_cw(state):
    """Prints out the single grid square state index for Cliff Walking (0-47)"""
    print(f"Current Grid Square Position Index: {state}")

print("\nInitial State:")
describe_state_cw(state_cw)

# Actions: 0 (Up), 1 (Right), 2 (Down), 3 (Left)
actions_cw = {0: "Move Up", 1: "Move Right", 2: "Move Down", 3: "Move Left"}
print("\nActions Available:")
for act in actions_cw:
    print(f"Action {act}: {actions_cw[act]}")

print("\nSimulating 2 steps:")
for step in range(2):
    action = env_cw.action_space.sample()
    next_state, reward, terminated, truncated, info = env_cw.step(action)
    done = terminated or truncated
    print(f"Step {step + 1}: Action taken -> {actions_cw[action]}")
    describe_state_cw(next_state)
    print(f"Reward: {reward} | Done: {done}")
env_cw.close()
```
