```python
import time

class SimpleRecyclingRobot:
    def __init__(self):
        # The robot starts at Home Base [0,0]
        self.current_x = 0
        self.current_y = 0
        self.gripper_empty = True
        self.carried_item = None
        
        # Fixed locations in our virtual room [X, Y]
        self.pickup_station = [5, 0]
        self.plastic_bin = [5, 5]
        self.paper_bin = [0, 5]

    def navigate_to(self, target_label, target_coords):
        """Moves the robot step-by-step toward a destination coordinate."""
        target_x, target_y = target_coords
        print(f"\n[NAV] Starting travel to {target_label} at [{target_x}, {target_y}]...")
        
        # Move along X axis
        while self.current_x != target_x:
            if self.current_x < target_x:
                self.current_x += 1
            else:
                self.current_x -= 1
            print(f"  > Driving... Current Position: [{self.current_x}, {self.current_y}]")
            time.sleep(0.4) # Pause to simulate physical movement

        # Move along Y axis
        while self.current_y != target_y:
            if self.current_y < target_y:
                self.current_y += 1
            else:
                self.current_y -= 1
            print(f"  > Driving... Current Position: [{self.current_x}, {self.current_y}]")
            time.sleep(0.4)
            
        print(f"[NAV] Arrived safely at {target_label}!")

    def vision_scan(self):
        """Simulates looking at an item to check what material it is."""
        print("\n[VISION] Camera scanning the conveyor belt...")
        time.sleep(1)
        
        # Let's pretend the camera found a plastic bottle
        detected_material = "plastic" 
        print(f"[VISION] Identified object material: {detected_material.upper()}")
        return detected_material

    def pick_up_item(self, item_name):
        """Controls the arm mechanism to grab the object."""
        print(f"\n[ARM] Lowering gripper over the {item_name}...")
        time.sleep(1)
        self.gripper_empty = False
        self.carried_item = item_name
        print("[ARM] Gripper secured! Lifting item up.")

    def place_in_bin(self):
        """Drops the item into the designated bin."""
        print(f"\n[ARM] Extending arm over the bin...")
        time.sleep(1)
        print(f"[ARM] Opening gripper. {self.carried_item.upper()} dropped into bin.")
        self.gripper_empty = True
        self.carried_item = None

    def run_complete_cycle(self):
        """The main instructions the robot executes from start to finish."""
        print("=== RECYCLING ROBOT MISSION STARTED ===")
        
        # Step 1: Drive to where the trash arrives
        self.navigate_to("Pickup Station", self.pickup_station)
        
        # Step 2: Look at the trash item
        item = self.vision_scan()
        
        # Step 3: Physically grab it
        self.pick_up_item(item)
        
        # Step 4: Decide which bin to go to based on vision result
        if item == "plastic":
            destination_bin = "Plastic Bin"
            destination_coords = self.plastic_bin
        elif item == "paper":
            destination_bin = "Paper Bin"
            destination_coords = self.paper_bin
        else:
            print("Unknown material! Staying put.")
            return

        # Step 5: Drive to the correct bin while holding the item
        self.navigate_to(destination_bin, destination_coords)
        
        # Step 6: Drop the item inside
        self.place_in_bin()
        
        # Step 7: Head back home to wait for more trash
        self.navigate_to("Home Base", [0, 0])
        print("\n=== MISSION SUCCESSFUL: READY FOR NEXT CYCLE ===")

# Run the program
if __name__ == "__main__":
    robot = SimpleRecyclingRobot()
    robot.run_complete_cycle()
```
