```python
import random
import pandas as pd

def run_bandit_experiment(file_path):
    # 1. Configuration parameters
    total_rounds = 10000
    d = 10  # Number of ads
    epsilon = 0.1
    
    # 2. Variable initializations using exact structural formatting
    number_of_selections = [0] * d
    sum_of_rewards = [0] * d
    
    # Track overall progression statistics
    total_reward_collected = 0
    
    # Load dataset containing click matrix history
    df = pd.read_csv(file_path)
    max_available_rows = len(df)
    
    # Safety fallback to prevent indexing past the file size layout bounds
    if total_rounds > max_available_rows:
        total_rounds = max_available_rows

    # 3. Main multi-armed bandit iteration timeline loop
    for round_idx in range(total_rounds):
        
        # Calculate standard average estimated CTR per ad to prepare choices
        estimated_ctrs = [0.0] * d
        for i in range(d):
            if number_of_selections[i] > 0:
                estimated_ctrs[i] = sum_of_rewards[i] / number_of_selections[i]
        
        # --- Epsilon-Greedy Action Selection Process ---
        if random.random() < epsilon:
            # Exploration Phase: Choose a completely random ad arm index
            selected_ad = random.randint(0, d - 1)
        else:
            # Exploitation Phase: Find the ad with the maximum average score conversion
            max_ctr_val = max(estimated_ctrs)
            
            # Identify all indices tied for top score to ensure fair non-biased evaluation
            best_ad_candidates = [i for i, val in enumerate(estimated_ctrs) if val == max_ctr_val]
            selected_ad = random.choice(best_ad_candidates)
            
        # 4. Fetch outcome reward token vector from column data row matching current step
        reward = df.iloc[round_idx, selected_ad]
        
        # 5. In-place history state tracking updates
        number_of_selections[selected_ad] += 1
        sum_of_rewards[selected_ad] += reward
        total_reward_collected += reward

    # Calculate final True CTR to determine the absolute winner
    final_ctrs = [0.0] * d
    for i in range(d):
        if number_of_selections[i] > 0:
            final_ctrs[i] = sum_of_rewards[i] / number_of_selections[i]

    # Find the ad with the highest Click-Through Rate (CTR)
    best_ad_index = final_ctrs.index(max(final_ctrs))

    # 6. Final report rendering logic output 
    print("================ EXPERIMENT METRICS ================")
    print(f"Total Simulation Rounds Simulated: {total_rounds}")
    print(f"Overall Click Count Yielded: {total_reward_collected}")
    print("\nIndividual Item Performance Logs:")
    
    for i in range(d):
        print(f"Ad Index {i:02d} | Display Count: {number_of_selections[i]:4d} | "
              f"Clicks Accumulated: {sum_of_rewards[i]:3d} | CTR: {final_ctrs[i]:.4f}")
              
    print("====================================================")
    print(f"★ THE BEST AD IS: Ad {best_ad_index} ★")
    print("====================================================")

# --- Execution Command ---
# This runs the function using your local file 'a.csv'
run_bandit_experiment('a.csv')
```
