```python
import gymnasium as gym
import numpy as np

# ------------------------------------------------------
# Step 1: Create the FrozenLake environment
# ------------------------------------------------------
env = gym.make("FrozenLake-v1", is_slippery=False)

# Print number of states and actions
print("Number of states:", env.observation_space.n)
print("Number of actions:", env.action_space.n)

# ------------------------------------------------------
# Step 2: Initialize variables
# ------------------------------------------------------
# Create a value function table filled with zeros
V = np.zeros(env.observation_space.n)

# Define parameters
gamma = 0.9  # Discount factor
theta = 1e-6 # Small threshold for convergence

# ------------------------------------------------------
# Step 3: Value Iteration Algorithm
# ------------------------------------------------------
while True:
    delta = 0  # Tracks the maximum change in value function
    
    # Loop through each state
    for s in range(env.observation_space.n):
        # Store the old value of state 's'
        v = V[s]
        
        # Initialize a list to store action values for all actions
        action_values = []
        
        # Loop through all possible actions
        for a in range(env.action_space.n):
            # Initialize value for this action
            action_value = 0
            
            # Use env.unwrapped.P to look up dynamics
            for prob, next_state, reward, done in env.unwrapped.P[s][a]:
                # Bellman equation for value update
                action_value += prob * (reward + gamma * V[next_state])
                
            # Append this action value
            action_values.append(action_value)
            
        # Update the state value with the best action value (max)
        V[s] = max(action_values)
        
        # Update delta to check convergence
        delta = max(delta, abs(v - V[s]))
        
    # If values have converged, break
    if delta < theta:
        break

# ------------------------------------------------------
# Step 4: Extract Optimal Policy
# ------------------------------------------------------
policy = np.zeros(env.observation_space.n, dtype=int)

# For each state, choose the best action
for s in range(env.observation_space.n):
    action_values = np.zeros(env.action_space.n)
    for a in range(env.action_space.n):
        for prob, next_state, reward, done in env.unwrapped.P[s][a]:
            action_values[a] += prob * (reward + gamma * V[next_state])
            
    # Pick the action with maximum value
    policy[s] = np.argmax(action_values)

# ------------------------------------------------------
# Step 5: Print Results
# ------------------------------------------------------
print("\nOptimal Value Function:")
print(V)

print("\nOptimal Policy (action per state):")
print(policy.reshape(4, 4))
```
